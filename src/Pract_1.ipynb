### Аналіз швидкості усиновлення

*0 означає, що тварину усиновили в день, коли вона була додана.
1 означає, що тварину усиновили протягом 1-7 днів.
2 означає, що тварину усиновили протягом 8-30 днів.
3 означає, що тварину усиновили протягом 31-90 днів.
4 означає, що тварина не була усиновлена або усиновлена після 100+ днів.*

```python
plt.figure(figsize=(5, 7))

palette = sns.color_palette("viridis", as_cmap=False)
g = sns.countplot(x='AdoptionSpeed', data=ds, palette=palette)

plt.title('Відсоткове співвідношення швидкості усиновлення')

ax = g.axes
total = ds.shape[0]  # Загальна кількість записів
for p in ax.patches:
    # Обчислюємо відсоток
    percentage = (p.get_height() * 100 / total)
    # Додаємо анотацію з відсотком
    ax.annotate(f"{percentage:.2f}%", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', fontsize=11, color='gray', rotation=0, xytext=(0, 10),
                textcoords='offset points')

plt.show()
```

```python
ds['Type'] = ds['Type'].map({1: 'Собаки', 2: 'Коти'})

plt.figure(figsize=(15, 8))

g = sns.countplot(x='Type', data=ds, hue='AdoptionSpeed')

plt.title('Швидкість усиновлення за типом тварини')

ax = g.axes

for p in ax.patches:
    height = p.get_height()
    x = p.get_x() + p.get_width() / 2
    hue = p.get_label()
    
    if height > 0:
        text = f"{int(height)}"
    else:
        text = '0'
    color = 'blue' 
    ax.annotate(text, (x, height),
                ha='center', va='center', fontsize=11, color=color,
                rotation=0, xytext=(0, 10), textcoords='offset points')

plt.show()
```

```python
# Ділимо вік на групи
bins = [0, 6, 12, 24, 36, 48, 60, 72, 84, 96, 108, 120]
labels = ['0-6', '7-12', '13-24', '25-36', '37-48', '49-60', '61-72', '73-84', '85-96', '97-108', '109-120']
ds['AgeGroup'] = pd.cut(ds['Age'], bins=bins, labels=labels)

# Обчислюємо середню швидкість усиновлення по вікових групах
avg_adoption_speed = ds.groupby('AgeGroup')['AdoptionSpeed'].mean().reset_index()

# Створюємо графік Line Plot
plt.figure(figsize=(12, 8))
plt.plot(avg_adoption_speed['AgeGroup'], avg_adoption_speed['AdoptionSpeed'], marker='o')
plt.xlabel('Вікова група (місяці)')
plt.ylabel('Середня швидкість усиновлення')
plt.title('Середня швидкість усиновлення за віковими групами')
plt.grid(True)

# Показуємо графік
plt.show()
```

```python

# Створимо функцію для побудови графіків
def plot_histogram(ds, column, position):
    plt.subplot(5, 3, position)  # 5 рядів, 3 стовпці
    sns.histplot(data=ds, x=column, hue='AdoptionSpeed', multiple='stack', palette='YlOrRd', kde=True)
    plt.title(f"Plot for '{column}'")
    plt.xlabel(column)
    plt.ylabel('Count')

# Створимо графіки
plt.figure(figsize=(15, 25))

# Виклик функції для кожної колонки
columns = ['Type', 'Age', 'Gender', 'MaturitySize', 'FurLength','Health',  'Sterilized', 'Vaccinated', 'Dewormed', 'VideoAmt', 
           'PhotoAmt',   'Fee', 'State', 'Quantity']

# Побудова графіків для кожної колонки
for i, col in enumerate(columns):
    plot_histogram(ds, col, i+1)  # передаємо позицію у subplot

# Показати графіки
plt.tight_layout()
plt.show()
```

```python
def make_count_plot(df, x, hue, title):
    plt.figure(figsize=(10, 4))
    sns.countplot(x=x, hue=hue, data=df, palette='YlOrRd')
    plt.title(title, fontsize=16)
    plt.xlabel(x.capitalize(), fontsize=14)
    plt.ylabel('Кількість тварин', fontsize=14)
    plt.xticks(rotation=45, ha='right')
    plt.legend(title=hue.capitalize())
    plt.tight_layout()
    plt.show()

make_count_plot(df=ds, x='Gender', hue='AdoptionSpeed', title='Кількість тварин за гендером і швидкістю усиновлення')
```

```python
# Імпортуємо необхідні бібліотеки
import pandas as pd
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Завантажуємо дані
ds = pd.read_csv("variant_3.csv")

# Фільтруємо тільки котів (Type = 2)
cats = ds[ds['Type'] == 2]

# Заповнюємо відсутні імена пустими рядками
cats['Name'].fillna('', inplace=True)

# Функція для поділу імен
def split_names(name):
    # Якщо ім'я містить &, поділяємо на окремі імена
    if '&' in name:
        return name.split('&')
    else:
        return [name]

# Створюємо список імен котів, ігноруючи порожні і "UNNAMED" імена
names_list = []
for name in cats['Name']:
    if name.strip().lower() != 'unnamed' and name.strip() != '':  # Ігноруємо порожні та "UNNAMED" імена
        names_list.extend(split_names(name))

# Об'єднуємо всі імена в один текстовий рядок
text = ' '.join(name.strip() for name in names_list)

# Генеруємо хмару слів
wordcloud = WordCloud(width=800, height=400, background_color='white', colormap='plasma').generate(text)

# Виводимо хмару слів
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Видаляємо осі
plt.show()
```

```python
# Фільтруємо тільки собак (Type = 1)
dogs = ds[ds['Type'] == 1]

# Заповнюємо відсутні імена пустими рядками
dogs['Name'].fillna('', inplace=True)

# Створюємо список імен собак, ігноруючи порожні і "UNNAMED" імена
names_list = []
for name in dogs['Name']:
    if name.strip().lower() != 'unnamed' and name.strip() != '':  # Ігноруємо порожні та "UNNAMED" імена
        names_list.extend(split_names(name))

# Об'єднуємо всі імена в один текстовий рядок
text = ' '.join(name.strip() for name in names_list)

# Генеруємо хмару слів
wordcloud = WordCloud(width=800, height=400, background_color='white', colormap='plasma').generate(text)

# Виводимо хмару слів
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Видаляємо осі
plt.show()
```
